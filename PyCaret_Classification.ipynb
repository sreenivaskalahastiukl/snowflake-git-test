```python
# loading the dataset
from pycaret.datasets import get_data
dataset = get_data('credit')
```

```python
# check the shape of data
dataset.shape
```

```python
# sample 5% of data to be used as unseen data
data = dataset.sample(frac=0.95, random_state=786)
data_unseen = dataset.drop(data.index)
data.reset_index(inplace=True, drop=True)
data_unseen.reset_index(inplace=True, drop=True)
# print the revised shape
print('Data for Modeling: ' + str(data.shape))
print('Unseen Data For Predictions: ' + str(data_unseen.shape))
```

```python
# init setup
from pycaret.classification import *
s = setup(data = data, target = 'default', session_id=123)
```

```python
#Comparing All Models
best_model = compare_models()
```

```python
print(best_model)
```

```python
# Create a Model
# check available models
models()
```

```python
#Decision Tree Classifier
dt = create_model('dt')
```

```python
# trained model object is stored in the variable 'dt'.
print(dt)
```

```python
#K Neighbors Classifier
knn = create_model('knn')
```

```python
# Random Forest Classifier
rf = create_model('rf')
```

```python
#Tune a Model
#Decision Tree Classifier
tuned_dt = tune_model(dt)
```

```python
# tuned model object is stored in the variable 'tuned_dt'.
print(tuned_dt)
```

```python
#K Neighbors Classifier
import numpy as np
tuned_knn = tune_model(knn, custom_grid = {'n_neighbors' : np.arange(0,50,1)})
```

```python
print(tuned_knn)
```

```python
#Random Forest Classifier
tuned_rf = tune_model(rf)
```

```python
#Plot a Model
#AUC Plot
plot_model(tuned_rf, plot = 'auc')
```

```python
#Precision-Recall Curve
plot_model(tuned_rf, plot = 'pr')
```

```python
#Feature Importance Plot
plot_model(tuned_rf, plot='feature')
```

```python
#Confusion Matrix
plot_model(tuned_rf, plot = 'confusion_matrix')
```

```python
evaluate_model(tuned_rf)
```

```python
#Predict on test / hold-out Sample
predict_model(tuned_rf);
```

```python
#Finalize Model for Deployment
# finalize rf model
final_rf = finalize_model(tuned_rf)
# print final model parameters
print(final_rf)
```

```python
predict_model(final_rf);
```

```python
#Predict on unseen data
unseen_predictions = predict_model(final_rf, data=data_unseen)
unseen_predictions.head()
```

```python
# check metric on unseen data
from pycaret.utils import check_metric
check_metric(unseen_predictions['default'], unseen_predictions['Label'], metric = 'Accuracy')
```

```python
#Saving the model
#Saving the final model
save_model(final_rf,'Final RF Model 02Sep2023')
```

```python
#Loading the saved model
# loading the saved model
saved_final_rf = load_model('Final RF Model 02Sep2023')
```

```python
# predict on new data
new_prediction = predict_model(saved_final_rf, data=data_unseen)
new_prediction.head()
```

```python
from pycaret.utils import check_metric
check_metric(new_prediction['default'], new_prediction['Label'], metric = 'Accuracy')
```
